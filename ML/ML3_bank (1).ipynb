```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.metrics import classification_report
```

```python

df = pd.read_csv("Churn_Modelling.csv")

print(df.head())
print(df.shape)
```

```python

# Separate features and target

X = df.drop(
    ["RowNumber", "CustomerId", "Surname", "Exited"],
    axis=1
)

y = df["Exited"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

```python

# Convert categorical data into numerical data

X = pd.get_dummies(
    X,
    columns=["Geography", "Gender"],
    drop_first=True,
    dtype=int
)

print(X.head())
print("\nFeature shape:", X.shape)
```

```python

# Split the dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)

```

```python

# Normalize the data

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("Training data normalized successfully.")
print("Testing data normalized successfully.")

print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python

# Build the Neural Network model

model = MLPClassifier(
    hidden_layer_sizes=(32, 16),
    activation="relu",
    solver="adam",
    max_iter=500,
    random_state=42,
    early_stopping=True
)

# Train the model
model.fit(X_train, y_train)

print("Neural Network model trained successfully!")
```

```python

# Predict customer churn

y_pred = model.predict(X_test)

print("Actual values:")
print(y_test.head(10).to_numpy())

print("\nPredicted values:")
print(y_pred[:10])
```

```python

# Calculate accuracy

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy Score:", accuracy)
print("Accuracy Percentage:", round(accuracy * 100, 2), "%")
```

```python

# Calculate confusion matrix

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python

# Print classification report

print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python

print("Bank Customer Churn Prediction using Neural Network")
print("---------------------------------------------------")
print("Accuracy:", round(accuracy * 100, 2), "%")
print("\nConfusion Matrix:")
print(cm)
print("\nModel evaluation completed successfully.")
```
